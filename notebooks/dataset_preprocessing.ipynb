```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
for candidate in [PROJECT_ROOT, *PROJECT_ROOT.parents]:
    if (candidate / "pyproject.toml").exists():
        PROJECT_ROOT = candidate
        break
sys.path.insert(0, str(PROJECT_ROOT / "src"))

VERSIONS = ["v1.0", "v1.1", "v1.2", "v1.3"]

CONFIG_DIR = PROJECT_ROOT / "scripts" / "data" / "preprocessing" / "iu-xray"
for version in VERSIONS:
    assert (CONFIG_DIR / f"config_iu-xray_{version}.yaml").exists(), version
print("Config trovati per:", VERSIONS)
```

```python
import yaml

from shield.data.preprocessing import run_preprocessing

for version in VERSIONS:
    config_path = CONFIG_DIR / f"config_iu-xray_{version}.yaml"
    config = yaml.safe_load(config_path.read_text(encoding="utf-8"))
    out_dir = PROJECT_ROOT / "dataset" / "iu-xray" / f"iu-xray_{version}"
    if out_dir.exists():
        print(f"[skip] {out_dir.name} esiste già.")
        continue
    print(f"\n===== {version} =====")
    result = run_preprocessing(config, PROJECT_ROOT)
    print(f"[ok] output: {result['output_dir']} | esempi: {result['n_examples']} | drop immagini: {result['n_dropped_missing_image']}")
```
